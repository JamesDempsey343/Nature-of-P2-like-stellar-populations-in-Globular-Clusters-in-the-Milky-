```python
def check_columns(file):
    f = fits.open(file)
    print(f"\nChecking: {file}\n")

    for name in f[1].data.columns.names:
        arr = f[1].data[name]
        if arr.ndim == 1:
            print(f"{name:25s}  YES (1-D)")
        else:
            print(f"{name:25s}  NO (1-D)  shape={arr.shape}")
    f.close()

check_columns(members)
check_columns(params)
```

```python
print("\n[Mg/H] vs [Al/Mg] abundance completeness:\n")
print(f"{'Cluster':<10} {'Type':<6} {'Total':<8} {'Finite Mg/Al/Fe':<18}")

for cluster_name in np.unique(gc_names):

    cluster_mask = (gc_names == cluster_name)
    total_members = np.sum(cluster_mask)

    finite_mask = cluster_mask & \
                  np.isfinite(data["MG_FE"]) & \
                  np.isfinite(data["AL_FE"]) & \
                  np.isfinite(data["FE_H"])

    usable_members = np.sum(finite_mask)
    cluster_type = cluster_type_map[cluster_name]

    print(f"{cluster_name:<10} {cluster_type:<6} {total_members:<8} {usable_members:<18}")
```

```python
members = "~/Downloads/GC_members_VAC-v1_1.fits"
prob_csv = "~/Downloads/201125_wcen_ids_and_probs.csv"

with fits.open(members) as hdul:
    table = hdul[1].data
    cols = {}

    for name in table.columns.names:
        arr = np.array(table[name])
        if arr.ndim == 1:
            if arr.dtype.byteorder == ">":
                arr = arr.byteswap().view(arr.dtype.newbyteorder("="))
            cols[name] = arr

plot_data = pd.DataFrame(cols)
probs = pd.read_csv(prob_csv)

plot_data["GC_NAME"] = plot_data["GC_NAME"].astype(str).str.strip()
plot_data["APOGEE_ID"] = plot_data["APOGEE_ID"].astype(str).str.strip()
probs["APOGEE_ID"] = probs["APOGEE_ID"].astype(str).str.strip()
probs["flag"] = probs["flag"].astype(str).str.strip()

probs = probs.sort_values("prob", ascending=False)
probs = probs.drop_duplicates(subset="APOGEE_ID", keep="first").copy()

for col in ["LOGG", "FE_H", "MG_FE", "AL_FE"]:
    plot_data[col] = pd.to_numeric(plot_data[col], errors="coerce")

omega = plot_data[(plot_data["GC_NAME"] == "NGC5139") & (plot_data["LOGG"] < 3.6)].copy()
omega = omega.drop_duplicates(subset="APOGEE_ID")

omega["MG_H"] = omega["MG_FE"] + omega["FE_H"]
omega["AL_MG"] = omega["AL_FE"] - omega["MG_FE"]

omega = omega.merge(probs[["APOGEE_ID", "flag", "prob"]], on="APOGEE_ID", how="left")
omega["flag"] = omega["flag"].where(omega["flag"].isin(["P1", "IM", "P2"]), "Unlabeled")

omega_plot_df = omega[np.isfinite(omega["MG_H"]) & np.isfinite(omega["AL_MG"])].copy()

print("Total plotted points:", len(omega_plot_df))
print(omega_plot_df["flag"].value_counts())

bg_colors = {
    "Unlabeled": "#8c8c8c",
    "P1": "#b65a73",
    "P2": "#d5ab34",
    "IM": "#1f5aa6"
}

bg_sizes = {
    "Unlabeled": 30,
    "P1": 40,
    "P2": 40,
    "IM": 40
}

fig, ax = plt.subplots(figsize=(7, 6))

for group in ["Unlabeled", "P1", "P2", "IM"]:
    sub = omega_plot_df[omega_plot_df["flag"] == group]
    ax.scatter(
        sub["MG_H"],
        sub["AL_MG"],
        s=bg_sizes[group],
        c=bg_colors[group],
        alpha=1.0,
        edgecolors="none",
        label=group
    )

ax.axhline(0.0, color="black", linestyle=":", linewidth=1.2)

ax.set_xlabel("[Mg/H]", fontsize=15)
ax.set_ylabel("[Al/Mg]", fontsize=15)
ax.set_xlim(-2.5, 0.0)
ax.set_ylim(-1.2, 1.6)

ax.set_xticks(np.arange(-2.4, 0.01, 0.4))
ax.set_yticks(np.arange(-1.2, 1.61, 0.4))
ax.xaxis.set_minor_locator(MultipleLocator(0.2))
ax.yaxis.set_minor_locator(MultipleLocator(0.2))

ax.tick_params(axis="both", which="major", labelsize=12, direction="in", top=True, right=True, length=6)
ax.tick_params(axis="both", which="minor", direction="in", top=True, right=True, length=3)

ax.legend(frameon=False, fontsize=12, loc="upper right")

plt.tight_layout()
plt.show()
```

```python
omega_name = "NGC5139"

unique_clusters, counts = np.unique(gc_names, return_counts=True)
valid_clusters = unique_clusters[counts >= 10]

omega_mask = (data["GC_NAME"] == omega_name) & (data["LOGG"] < 3.6)

omega_ids   = np.array(data["APOGEE_ID"][omega_mask]).astype(str)
omega_feh   = np.array(data["FE_H"][omega_mask], dtype=float)
omega_mgfe  = np.array(data["MG_FE"][omega_mask], dtype=float)
omega_alfe  = np.array(data["AL_FE"][omega_mask], dtype=float)

omega_base = pd.DataFrame({
    "APOGEE_ID": omega_ids,
    "MG_H": omega_mgfe + omega_feh,
    "AL_MG": omega_alfe - omega_mgfe
})

omega_base["APOGEE_ID"] = omega_base["APOGEE_ID"].str.strip()
omega_base = omega_base.drop_duplicates(subset="APOGEE_ID")

probs = pd.read_csv("~/Downloads/201125_wcen_ids_and_probs.csv")
probs["APOGEE_ID"] = probs["APOGEE_ID"].astype(str).str.strip()
probs["flag"] = probs["flag"].astype(str).str.strip()
probs = probs.sort_values("prob", ascending=False)
probs = probs.drop_duplicates(subset="APOGEE_ID", keep="first").copy()

omega_base = omega_base.merge(
    probs[["APOGEE_ID", "flag", "prob"]],
    on="APOGEE_ID",
    how="left"
)

omega_base["flag"] = omega_base["flag"].where(
    omega_base["flag"].isin(["P1", "IM", "P2"]),
    "Unlabeled"
)

omega_plot_df = omega_base[
    np.isfinite(omega_base["MG_H"]) & np.isfinite(omega_base["AL_MG"])
].copy()

bg_colors = {
    "Unlabeled": "#8c8c8c",
    "P1": "#b65a73",
    "P2": "#d5ab34",
    "IM": "#1f5aa6"
}

bg_sizes = {
    "Unlabeled": 30,
    "P1": 40,
    "P2": 40,
    "IM": 40
}

type_labels = ["Type 1", "Type 2", "Unidentified"]

print(f"\nOverlay comparison using Omega Cen (N={omega_plot_df.shape[0]})")

for cluster_type in type_labels:

    print(f"\nPlotting {cluster_type} clusters (>=10 stars)")

    for cluster_name in valid_clusters:

        if cluster_name == omega_name:
            continue

        if cluster_type_map[cluster_name] != cluster_type:
            continue

        cluster_mask = (gc_names == cluster_name)
        N_cluster = np.sum(cluster_mask)

        if N_cluster < 10:
            continue

        cluster_Mg_H = data["MG_FE"][cluster_mask] + data["FE_H"][cluster_mask]
        cluster_Al_Mg = data["AL_FE"][cluster_mask] - data["MG_FE"][cluster_mask]

        finite = np.isfinite(cluster_Mg_H) & np.isfinite(cluster_Al_Mg)
        cluster_Mg_H = cluster_Mg_H[finite]
        cluster_Al_Mg = cluster_Al_Mg[finite]

        fig, ax = plt.subplots(figsize=(7, 6))

        for group in ["Unlabeled", "P1", "P2", "IM"]:
            sub = omega_plot_df[omega_plot_df["flag"] == group]
            ax.scatter(
                sub["MG_H"],
                sub["AL_MG"],
                s=bg_sizes[group],
                c=bg_colors[group],
                alpha=1.0,
                edgecolors="none",
                label=group
            )

        ax.scatter(
            cluster_Mg_H,
            cluster_Al_Mg,
            s=50,
            color="lime",
            alpha=1.0,
            label=f"{cluster_name}\n (N={len(cluster_Mg_H)})"
        )

        ax.axhline(0.0, color="black", linestyle=":", linewidth=1.2)

        ax.set_xlabel("[Mg/H]", fontsize=15)
        ax.set_ylabel("[Al/Mg]", fontsize=15)
        x_all = np.concatenate([omega_plot_df["MG_H"].values, np.asarray(cluster_Mg_H)])
        y_all = np.concatenate([omega_plot_df["AL_MG"].values, np.asarray(cluster_Al_Mg)])

        x_min, x_max = np.nanmin(x_all), np.nanmax(x_all)
        y_min, y_max = np.nanmin(y_all), np.nanmax(y_all)

        x_pad = max(0.08, 0.08 * (x_max - x_min))
        y_pad = max(0.08, 0.08 * (y_max - y_min))

        ax.set_xlim(x_min - x_pad, x_max + x_pad)
        ax.set_ylim(y_min - y_pad, y_max + y_pad)

        ax.xaxis.set_major_locator(MultipleLocator(0.4))
        ax.yaxis.set_major_locator(MultipleLocator(0.4))
        ax.xaxis.set_minor_locator(MultipleLocator(0.2))
        ax.yaxis.set_minor_locator(MultipleLocator(0.2))

        ax.tick_params(axis="both", which="major", labelsize=12, direction="in", top=True, right=True, length=6)
        ax.tick_params(axis="both", which="minor", direction="in", top=True, right=True, length=3)

        ax.set_title(
            f"[Al/Mg] vs [Mg/H]: {cluster_name} ({cluster_type}) over Omega Cen",
            fontsize=14
        )

        legend = ax.legend(frameon=True, fontsize=9, loc="upper right")
        legend.get_frame().set_edgecolor("black")
        legend.get_frame().set_linewidth(1.2)
        plt.tight_layout()
        plt.show()
```
